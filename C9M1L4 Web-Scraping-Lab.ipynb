## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
import requests
from bs4 import BeautifulSoup
import pandas as pd
import json
```

Download the webpage at the url


```python
#your code goes here
response = requests.get(url)
```

Create a soup object


```python
soup = BeautifulSoup(response.content,'html.parser')
```

```python
soup_pc = soup.prettify()
print(soup_pc)
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
df = pd.DataFrame(columns=['No.','Programming Language', 'Created By', 'Average Salary', 'Learning Difficulty'])
rows=soup.find_all('tr')
#type(rows)
for i, row in enumerate(rows):
    #print(row.find_all('td'))
    #print(f"Row {i}: {row.text}")
    if i==0:
        continue
    for j, col in enumerate(row.find_all('td')):
        if j==1:
            df.loc[i,'Programming Language'] = col.text
        if j==3:
            df.loc[i,'Average Salary'] = col.text
        if j==0:
            df.loc[i,'No.'] = col.text
        if j==2:
            df.loc[i,'Created By'] = col.text
df.head()
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
df[['Programming Language','Average Salary']].to_csv('popular-languages.csv', index=False)
```
