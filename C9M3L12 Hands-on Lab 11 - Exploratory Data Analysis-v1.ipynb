#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
# df['YearsCodePro'].replace({"Less than 1 year":"0"},inplace=True)
# df['YearsCodePro'].replace({"More than 50 years":"51"},inplace=True)
# df['YearsCodePro'] = df['YearsCodePro'].replace({"Less than 1 year":"0"})
# df['YearsCodePro'] = df['YearsCodePro'].replace({"More than 50 years":"51"})
# df.replace({'YearsCodePro': {"Less than 1 year": "0"}}, inplace=True)
# df.replace({'YearsCodePro': {"More than 50 years": "51"}}, inplace=True)
df.replace({'YearsCodePro': {"Less than 1 year": "0", "More than 50 years": "51"}}, inplace=True)
df['YearsCodePro'] = df['YearsCodePro'].astype(float)
df['YearsCodePro']
# df_ycp = df['YearsCodePro'].replace({"Less than 1 year":"0"})
# df_ycp.replace({"More than 50 years":"51"},inplace=True)
# df_ycp = df_ycp.astype(float)
# df_ycp[df_ycp > 50].value_counts()
```

```python
import numpy as np
conditions = [(df['YearsCodePro'] >=0) & (df['YearsCodePro'] < 5),
              (df['YearsCodePro'] >=5) & (df['YearsCodePro'] < 10),
              (df['YearsCodePro'] >=10) & (df['YearsCodePro'] < 20),
              (df['YearsCodePro'] >=20)              
              ]
categories = ["Years0_5","Years5_10","Years10_20","Years_Above20"]
years_categories = np.select(conditions, categories, default="Other")
# len(years_categories[years_categories == "Other"])
df['YearsCodePro'].isnull().sum()
years_categories
```

```python
df_ylevels = pd.get_dummies(years_categories, dtype=int)
df_ylevels
df_n = pd.concat([df, df_ylevels], axis=1)
df_n.shape
```

```python
#df_n.groupby(categories).size()
m05 = df_n.loc[df_n['Years0_5'] == 1, 'JobSat'].median()
m510 = df_n.loc[df_n['Years5_10'] == 1, 'JobSat'].median()
m1020 = df_n.loc[df_n['Years10_20'] == 1, 'JobSat'].median()
m20 = df_n.loc[df_n['Years_Above20'] == 1, 'JobSat'].median()
print(m05, m510, m1020, m20)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
x = ['0-5 years', '5-10 years', '10-20 years', 'Above 20 years']
y = [m05, m510, m1020, m20]
plt.figure(figsize=(10, 6))
sns.barplot(x=x, y=y)
plt.title('Median Job Satisfaction by Years of Professional Coding Experience')
plt.xlabel('Years of Professional Coding Experience')
plt.ylabel('Median Job Satisfaction')
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
df.loc[(df['Country']=='South Korea')&(df['RemoteWork']=='Remote'),['Country','RemoteWork','LanguageHaveWorkedWith']]
```

```python
%pip install country_converter
```

```python
import country_converter as coco
import pandas as pd

cc = coco.CountryConverter()
country_u = df['Country'].unique()
continent_l = cc.convert(names=country_u, to='continent')
country_continent = pd.Series(data=continent_l,index=country_u)
```

```python

df_copy = df.copy()
df_copy['Continent'] = df_copy['Country'].map(country_continent)
df_copy['Continent']
```

```python
langs = (
    df_copy.assign(Lang=df_copy['LanguageHaveWorkedWith'].str.split(';'))
    .explode('Lang')
    .groupby('Continent')['Lang']
    .value_counts()
)
```

```python
## Write your code here
asia_counts = langs['Asia']
africa_counts = langs['Africa']
europe_counts = langs['Europe']
america_counts = langs['America']
oceania_counts = langs['Oceania']
combined_data = {
    'Asia': asia_counts,
    'Africa': africa_counts,
    'Europe': europe_counts,
    'Americas': america_counts,
    'Oceania': oceania_counts
}
df_continents = pd.DataFrame.from_dict(combined_data, orient='index').fillna(0).astype(int)
df_continents
```

```python
df_continents_save = df_continents.copy()
lt = 0
for col in df_continents.columns:
    cnt = df_continents[col].sum()
    if cnt >= 2000:
        lt += 1
    else:
        df_continents = df_continents.drop(columns=[col])
print(f"Number of languages with 2000 or more occurrences: {lt}")
```

```python
%matplotlib inline
df_continents.plot(kind='bar', figsize=(12, 8))
plt.xlabel('Continent')
plt.ylabel('Number of Languages')
plt.title('Language Distribution by Continent')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
#df['Employment'].value_counts()
emp_items = df['Employment'].str.split(';').explode().unique()
emp_items
```

```python
edl_items = df['EdLevel'].str.split(';').explode().unique()
edl_items
```

```python
## Write your code here
emp_matrix = pd.DataFrame({emp: df['Employment'].str.contains(emp, regex=False, na=False) for emp in emp_items})
edl_matrix = pd.DataFrame({edl: df['EdLevel'].str.contains(edl, regex=False, na=False) for edl in edl_items})
result_df = emp_matrix.astype(int).T.dot(edl_matrix.astype(int))
result_df
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df_long = result_df.stack().reset_index()
df_long.columns = ['Y_label', 'X_label', 'Value']
df_long.dropna(inplace=True)
df_long.isnull().sum()
#df_long
```

```python
plt.figure(figsize=(10, 6))
scale_factor = 1
plt.scatter(
    x=df_long['X_label'],
    y=df_long['Y_label'],
    s=df_long['Value'] * scale_factor,
    alpha=0.6,
    c=df_long['Value'],
    cmap='viridis'
)

plt.colorbar(label='Values')
plt.xlabel('Education Level')
plt.ylabel('Employment')
plt.title('Bubble Plot of Education Level and Employment')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
