### 4. Encoding Categorical Variables


```python
df['Employment'].str.split(';').explode().value_counts()
```

```python
r = df["Employment"].str.extract(r"^(Employed|Independent|Not employed|Student)")[0]
r.value_counts()
```

<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
encoded_df = pd.get_dummies(r,prefix="Employment", dtype=int, ) 
encoded_df.head()
```

```python
df = pd.concat([df, encoded_df], axis=1)
df.columns
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
missing_values = df.isnull().sum()
missing_values[missing_values == missing_values.max()]
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
df.select_dtypes(include=['number'])
```

```python
df['ConvertedCompYearly'].isnull().sum()
```

```python
df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median(), inplace=True)
df['ConvertedCompYearly'].isnull().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
print(df['RemoteWork'].value_counts())
print(df['RemoteWork'].isnull().sum())
df['RemoteWork'].fillna(df['RemoteWork'].mode()[0], inplace=True)
print(df['RemoteWork'].value_counts())
print(df['RemoteWork'].isnull().sum())
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
dfc = df['ConvertedCompYearly']
dfc.value_counts()
des = dfc.describe()
Min = dfc.min()
Max = dfc.max()
dfc_m = dfc / (Max-Min)
dfc_m.value_counts()
```
